####1.2 Explain overfitting and why it is a problem in statistical learning. If you have a small dataset containing ten data points, should you prefer a simple model with one parameter or a complex model with ten parameters? Support your answer with an explanation.


Overfitting represents a fundamental challenge in statistical learning where a model captures the random noise inherent in the training data rather than the underlying distribution. This is a significant problem because it leads to excellent performance on training data but results in very poor generalization when the model is exposed to previously unseen datasets. When a model memorizes specific training examples instead of learning broader patterns, its practical utility is severely diminished.

With a small dataset containing only ten data points, a simple model with one parameter is highly preferable to a complex model with ten parameters. Selecting a model with ten parameters for ten data points would allow the model to perfectly fit every single point, essentially memorizing the dataset and resulting in extreme overfitting. A simpler model restricts the capacity to fit noise, which promotes much better generalization and provides a more robust representation of the true underlying trend.

####1.3 There are two commonly used approaches to avoid overfitting: regularization and cross-validation. Describe each one, and give one concrete example of a regularization technique used in modern machine learning and deep learning models.

Regularization refers to a broad set of techniques that introduce a penalty constraint to the model loss function during training to discourage excessive complexity. By penalizing large weights or complex parameter configurations, regularization prevents the model from fitting the training data too closely, which helps maintain a smoother decision boundary that generalizes better to unseen instances.

Cross-validation is a robust evaluation technique where the available training data is partitioned into multiple distinct subsets - called folds - to ensure the model performance is consistent. In a standard k-fold cross-validation setup, the model is trained on a majority of the subsets and validated on the remaining portion, and this process is repeated k times with different splits to provide a reliable estimate of model performance on truly novel data.

A highly common and practical regularization technique in modern deep learning is **Dropout**, which randomly deactivates a pre-determined percentage of neurons during each training step. This process forces the network to learn redundant representations and prevents individual neurons from co-adapting too closely to specific training samples, thereby substantially reducing the risk of overfitting.

####1.4 Provide two examples of metrics used to evaluate the performance of Classification and regression models, and give a formula for each one. Give two examples of applications and appropriate metrics for each case.

To evaluate a classification model, one can use the **F1-Score**, which harmonic mean of precision and recall. The mathematical formula for this metric is written as twice the product of Precision and Recall divided by their sum, which is mathematically expressed as 2 * (Precision * Recall) / (Precision + Recall).

For regression tasks, a standard metric is the **Mean Squared Error**, which calculates the average of the squared differences between the predicted values and the actual observed values. The mathematical formula is represented as the sum of the squared differences between actual and predicted values for all data points divided by the total number of data points.

An application example for classification is email spam detection, where utilizing the F1-Score is highly appropriate because it balances the cost of missing spam messages against the high penalty of misclassifying important personal emails. For a regression application, predicting housing prices is a great example where using Mean Squared Error is appropriate because this metric heavily penalizes large estimation errors, which is critical when financial investments are at stake.

####1.5 Why are benchmarks useful in machine learning? Give two examples, including at least one modern benchmark used to evaluate deep learning or large language models.

Benchmarks are incredibly useful in machine learning because they provide standardized datasets and standardized evaluation protocols that allow researchers to compare different algorithms objectively. Without these shared reference points, it would be virtually impossible to determine whether a newly proposed model actually represents an advancement over existing methodologies.

A classic example of an early benchmark is the MNIST dataset, which consists of handwritten digits and has been used for decades to evaluate basic image recognition algorithms. A modern benchmark used to evaluate large language models is the Massive Multitask Language Understanding suite, commonly referred to as MMLU, which tests models across dozens of academic and professional subjects to measure general knowledge and reasoning capabilities.

####2.1 What is machine learning? Discuss its evolution over time, from classical statistical methods, through the deep learning revolution, to today's large-scale foundation models, and explain why it has become so widely adopted.

Machine learning is a subfield of computer science and artificial intelligence focused on building systems that learn from data to improve their performance without being explicitly programmed for every scenario. Instead of relying on rigid, hand-crafted rules, these systems analyze patterns in historical information to make predictions or decisions regarding new, unseen observations.

The field has evolved significantly from its early reliance on classical statistical methods such as linear regression and logistic regression, which required rigid mathematical assumptions and manual selection of features. During the early twenty-first century, the deep learning revolution transformed the landscape by introducing multi-layered artificial neural networks capable of automatically learning hierarchical representations directly from raw data, which eliminated much of the manual engineering effort. Today, the industry has transitioned into the era of large-scale foundation models, which are pre-trained on massive datasets and can be adapted to a vast range of downstream tasks with minimal additional training.

This rapid and widespread adoption is driven by the unprecedented growth in digital data collection, the exponential increase in computational power through specialized hardware, and the remarkable economic value these models generate by automating complex cognitive tasks across industries.

####2.3 What is the difference between classification and regression?

The primary difference between classification and regression lies in the nature of the target variable they are designed to predict. Classification focus on predicting discrete categorical labels, which means assigning input data into specific predefined classes or categories such as determining whether an email is spam or legitimate.

On the other hand, regression is concerned with predicting continuous numerical values, such as estimating the temperature, stock prices, or the future value of a financial asset based on historical trends.

####2.4 What is the difference between supervised learning, unsupervised learning, and reinforcement learning? Give one current real-world application of each.

Supervised learning involves training a model on a labeled dataset where each training sample is paired with its correct output label, allowing the model to learn the mapping from inputs to outputs directly. A prominent real-world application of supervised learning is medical image analysis, where algorithms are trained on labeled scans to detect anomalies such as tumors.

Unsupervised learning operates on unlabeled data where the goal is to discover hidden structures, groupings, or associations within the dataset without any external guidance or target variables. A common real-world application of this approach is customer segmentation, where retail companies group customers into distinct archetypes based on purchasing habits to design targeted marketing campaigns.

Reinforcement learning relies on an agent interacting with a dynamic environment to learn an optimal sequence of decisions through trial and error, guided by feedback in the form of rewards or penalties. A highly successful real-world application of reinforcement learning is autonomous driving, where self-driving vehicles learn how to navigate complex traffic scenarios safely by optimizing their path selection over time.

####2.5. What is deep learning, and how does it differ from traditional (“shallow”) machine learning in terms of feature engineering, data requirements, and computational cost? Give one example of a task where deep learning substantially outperforms traditional methods, and one where it does not

Deep learning is a specialized subset of machine learning based on artificial neural networks with many hidden layers, which are designed to mimic the complex processing structures of the human brain. This architecture allows deep learning models to automatically discover optimal feature representations directly from raw inputs, which contrasts sharply with traditional machine learning where domain experts must manually design and select features beforehand.

In terms of data requirements, traditional machine learning models can achieve high accuracy with relatively small datasets, whereas deep learning models typically require massive volumes of labeled data to train millions of internal parameters effectively. Additionally, the computational cost of deep learning is exceptionally high because training these multi-layered structures requires specialized hardware like graphics processing units, whereas traditional models can be trained quickly on standard central processing units with minimal resources.

An example of a task where deep learning substantially outperforms traditional methods is real-time language translation, where deep neural networks capture complex semantic nuances across long sentences. Conversely, a task where traditional methods often perform just as well or even better is predicting tabular business data, such as forecasting monthly sales from a small historical spreadsheet, where simple gradient boosted trees are often more efficient and interpretable.

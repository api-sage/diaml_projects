####1.2 Explain overfitting and why it is a problem in statistical learning. If you have a small dataset containing ten data points, should you prefer a simple model with one parameter or a complex model with ten parameters? Support your answer with an explanation.


Overfitting represents a fundamental challenge in statistical learning where a model captures the random noise inherent in the training data rather than the underlying distribution. This is a significant problem because it leads to excellent performance on training data but results in very poor generalization when the model is exposed to previously unseen datasets. When a model memorizes specific training examples instead of learning broader patterns, its practical utility is severely diminished.

With a small dataset containing only ten data points, a simple model with one parameter is highly preferable to a complex model with ten parameters. Selecting a model with ten parameters for ten data points would allow the model to perfectly fit every single point, essentially memorizing the dataset and resulting in extreme overfitting. A simpler model restricts the capacity to fit noise, which promotes much better generalization and provides a more robust representation of the true underlying trend.

####1.3 There are two commonly used approaches to avoid overfitting: regularization and cross-validation. Describe each one, and give one concrete example of a regularization technique used in modern machine learning and deep learning models.

Regularization refers to a broad set of techniques that introduce a penalty constraint to the model loss function during training to discourage excessive complexity. By penalizing large weights or complex parameter configurations, regularization prevents the model from fitting the training data too closely, which helps maintain a smoother decision boundary that generalizes better to unseen instances.

Cross-validation is a robust evaluation technique where the available training data is partitioned into multiple distinct subsets - called folds - to ensure the model performance is consistent. In a standard k-fold cross-validation setup, the model is trained on a majority of the subsets and validated on the remaining portion, and this process is repeated k times with different splits to provide a reliable estimate of model performance on truly novel data.

A highly common and practical regularization technique in modern deep learning is **Dropout**, which randomly deactivates a pre-determined percentage of neurons during each training step. This process forces the network to learn redundant representations and prevents individual neurons from co-adapting too closely to specific training samples, thereby substantially reducing the risk of overfitting.

####1.4 Provide two examples of metrics used to evaluate the performance of Classification and regression models, and give a formula for each one. Give two examples of applications and appropriate metrics for each case.

To evaluate a classification model, one can use the **F1-Score**, which harmonic mean of precision and recall. The mathematical formula for this metric is written as twice the product of Precision and Recall divided by their sum, which is mathematically expressed as 2 * (Precision * Recall) / (Precision + Recall).

For regression tasks, a standard metric is the **Mean Squared Error**, which calculates the average of the squared differences between the predicted values and the actual observed values. The mathematical formula is represented as the sum of the squared differences between actual and predicted values for all data points divided by the total number of data points.

An application example for classification is email spam detection, where utilizing the F1-Score is highly appropriate because it balances the cost of missing spam messages against the high penalty of misclassifying important personal emails. For a regression application, predicting housing prices is a great example where using Mean Squared Error is appropriate because this metric heavily penalizes large estimation errors, which is critical when financial investments are at stake.

####3.1 Load the dataset and report, for each of the five regions: the number of listings, and the mean, median, and standard deviation of price. Why do the mean and median disagree in some regions?

```python
import pandas as pd
import numpy as np
import re
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.api as sm
from sklearn.metrics import mean_squared_error
from sklearn.model_selection import train_test_split
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, confusion_matrix
from sklearn.metrics import precision_score, recall_score, f1_score, roc_auc_score
```

```python
airbnb_url = "https://raw.githubusercontent.com/api-sage/diaml_projects/refs/heads/main/project5/spain_airbnb.csv"
airbnb_raw = pd.read_csv(airbnb_url, low_memory=False)

# The price column is stored as text with a dollar sign and a comma, so it is
# turned into a plain number before any statistics can be calculated.
airbnb_raw["price"] = airbnb_raw["price"].replace(r"[\$,]", "", regex=True).astype(float)

region_summary = airbnb_raw.groupby("region")["price"].agg(
    number_of_listings="count",
    mean_price="mean",
    median_price="median",
    std_price="std",
).round(2)

region_summary
```

Each of the five regions holds several thousand listings, so every regional average is based on a large and reasonably reliable sample. In every region the mean price is noticeably higher than the median price. This happens because the price variable is strongly right-skewed: most listings are modestly priced, but a smaller number of very expensive, luxury listings pull the average upward without moving the middle value much. Mallorca shows the biggest gap between mean and median, which tells us that Mallorca contains the largest share of unusually expensive, high-end listings compared with the other regions. Barcelona and Madrid also show a sizeable gap for the same reason, while Seville and Valencia, which have somewhat lower average prices overall, show a smaller gap, meaning their listings are a little more evenly spread around the typical price. In short, whenever a few very large values exist in a dataset, the mean gets dragged toward them while the median, which simply reports the middle-ranked value, stays close to what a typical listing actually costs.

####3.2 Clean the data in four steps, reporting how many rows are affected by each:
(i) Drop listings with a missing price;
(ii) Parse the number out of the bathrooms_text column and use it to fill missing values in the bathrooms column;
(iii) Since the Bedrooms and beds columns are correlated, fill missing values of bedrooms using the corresponding values of beds, and vice versa, as a simple imputation rule;
(iv) Drop the remaining rows with missing values.
Compare this number of remaining rows with what you would have gotten by simply dropping all rows with any missing value from the start, without the above steps. Discuss why systematically handling missing values like this is important in machine learning.

```python
relevant_columns = [
    "price", "accommodates", "bedrooms", "bathrooms", "beds", "minimum_nights",
    "calculated_host_listings_count", "number_of_reviews", "room_type", "region",
]

rows_before_cleaning = len(airbnb_raw)

# For comparison: what if we had simply dropped every row with any missing
# value among the columns we actually plan to use, right from the start?
naive_drop = airbnb_raw.dropna(subset=relevant_columns)

# Step (i): drop listings with a missing price
airbnb_clean = airbnb_raw.dropna(subset=["price"]).copy()
rows_dropped_step1 = rows_before_cleaning - len(airbnb_clean)

# Step (ii): pull the leading number out of bathrooms_text (e.g. "1.5 baths")
# and use it to fill in missing values of the bathrooms column. A text that
# only says "Half-bath" is treated as 0.5 bathrooms.
def parse_bathrooms_text(text):
    if pd.isna(text):
        return np.nan
    text = str(text).lower()
    if "half-bath" in text:
        return 0.5
    match = re.search(r"(\d+(\.\d+)?)", text)
    return float(match.group(1)) if match else np.nan

airbnb_clean["bathrooms_from_text"] = airbnb_clean["bathrooms_text"].apply(parse_bathrooms_text)
bathrooms_missing_before = airbnb_clean["bathrooms"].isna().sum()
fill_mask = airbnb_clean["bathrooms"].isna() & airbnb_clean["bathrooms_from_text"].notna()
airbnb_clean.loc[fill_mask, "bathrooms"] = airbnb_clean.loc[fill_mask, "bathrooms_from_text"]
rows_filled_step2 = int(fill_mask.sum())

# Step (iii): bedrooms and beds are highly correlated, so a missing bedrooms
# value is filled with the listing's beds value, and a missing beds value is
# filled with the listing's bedrooms value.
fill_bedrooms = airbnb_clean["bedrooms"].isna() & airbnb_clean["beds"].notna()
airbnb_clean.loc[fill_bedrooms, "bedrooms"] = airbnb_clean.loc[fill_bedrooms, "beds"]
fill_beds = airbnb_clean["beds"].isna() & airbnb_clean["bedrooms"].notna()
airbnb_clean.loc[fill_beds, "beds"] = airbnb_clean.loc[fill_beds, "bedrooms"]
rows_filled_step3 = int(fill_bedrooms.sum() + fill_beds.sum())

# Step (iv): drop whatever rows still have a missing value in the columns we need
rows_before_step4 = len(airbnb_clean)
airbnb_clean = airbnb_clean.dropna(subset=relevant_columns)
rows_dropped_step4 = rows_before_step4 - len(airbnb_clean)

print(f"Step (i)  - rows dropped for missing price:              {rows_dropped_step1}")
print(f"Step (ii) - bathrooms values filled from bathrooms_text:  {rows_filled_step2} (out of {bathrooms_missing_before} originally missing)")
print(f"Step (iii)- bedrooms/beds values filled from each other:  {rows_filled_step3}")
print(f"Step (iv) - rows dropped for remaining missing values:    {rows_dropped_step4}")
print(f"\nRows remaining after systematic cleaning:  {len(airbnb_clean)}")
print(f"Rows remaining after naive, one-shot dropna: {len(naive_drop)}")
```

Dropping listings with a missing price removes about ten thousand rows, since price is the value we are trying to predict and a model cannot be trained or evaluated on a row that has no answer to learn from. Parsing the bathrooms_text column recovers the bathrooms count for almost all of the listings that were missing it directly, because the two columns describe the same thing in two different formats, and one of them is usually present even when the other is not. Filling bedrooms from beds and beds from bedrooms similarly recovers a large number of rows, since a listing that specifies one of these two closely related counts almost always allows a reasonable guess at the other. After these three repair steps, only a small number of rows still have missing values and have to be dropped.

Comparing the two approaches shows the systematic method keeps thousands more rows than simply deleting every row that has any missing value at all. This matters a great deal in machine learning because deleting data carelessly can throw away a large fraction of the available information, shrink the training set, and bias the model if the missing values are not randomly distributed across listings. By first using information we already have in other columns to intelligently reconstruct missing values before resorting to deletion, we keep a larger and more representative dataset, which generally leads to a model that is trained on more examples and that generalizes better to new, unseen listings.

####3.3 Select the following explanatory variables to predict price: accommodates, bedrooms, bathrooms, beds, minimum_nights, calculated_host_listings_count, number_of_reviews, room_type, and region. Produce a correlation matrix of the seven numeric explanatory variables plus the dummy variables from categorical variables, make a heat map of the matrix, and describe the relationships between the variables.

```python
numeric_predictors = [
    "accommodates", "bedrooms", "bathrooms", "beds", "minimum_nights",
    "calculated_host_listings_count", "number_of_reviews",
]

# room_type and region are categories, so they are converted into 0/1 dummy
# variables (one category is left out of each group as the baseline/reference).
room_type_dummies = pd.get_dummies(airbnb_clean["room_type"], prefix="room", drop_first=True)
region_dummies = pd.get_dummies(airbnb_clean["region"], prefix="region", drop_first=True)

predictor_matrix = pd.concat(
    [airbnb_clean[numeric_predictors], room_type_dummies, region_dummies], axis=1
).astype(float)

correlation_matrix = predictor_matrix.corr()

plt.figure(figsize=(11, 9))
sns.heatmap(correlation_matrix, annot=True, fmt=".2f", cmap="coolwarm", center=0, square=True)
plt.title("Correlation matrix of Airbnb price predictors")
plt.tight_layout()
plt.show()
```

The heat map shows that accommodates, bedrooms, bathrooms, and beds move strongly together, which makes intuitive sense since a bigger home naturally has more sleeping capacity, more bedrooms, more beds, and more bathrooms. This is the strongest set of relationships in the matrix. The room type dummy variable for a private room is clearly and negatively related to accommodates, bedrooms, bathrooms, and beds, which simply reflects that a listing for a single private room within someone's home is, by definition, smaller than an entire home or apartment. Minimum nights, calculated host listings count, and number of reviews show only weak relationships with the other variables and with each other, meaning they each carry fairly independent information about a listing. The region dummy variables are also only weakly related to the size variables, except that Mallorca is mildly positively related to accommodates and the other size variables, suggesting that listings in Mallorca tend to be a little larger on average than listings elsewhere. None of the correlations between the different explanatory variables is so extreme that we must worry about severe multicollinearity in the regression model, aside from the expected relationship among the size-related variables.

####3.4 Create a multivariate linear model (model1) using the selected variables and a constant. What are the Mean Squared Error and the adjusted R2 for the model? Are all variables significant?

```python
X_full = sm.add_constant(predictor_matrix)
y_price = airbnb_clean["price"].astype(float)

model1 = sm.OLS(y_price, X_full).fit()
print(model1.summary())

model1_predictions = model1.predict(X_full)
model1_mse = mean_squared_error(y_price, model1_predictions)
print(f"\nMean Squared Error (in-sample): {model1_mse:,.2f}")
print(f"Adjusted R-squared:             {model1.rsquared_adj:.4f}")
```

Model1's in-sample Mean Squared Error is very large, which matches what we already saw in question 3.1: the price variable has an enormous range because a small number of extremely expensive listings sit far above the typical price, and squaring those large errors inflates the Mean Squared Error substantially. The adjusted R-squared is modest, around a quarter, meaning the selected variables together explain only about one quarter of the variation in price once we account for the number of predictors used; most of the variation in price is driven by factors not captured in this simple model, such as the quality of the listing, its exact neighbourhood, or seasonal demand.

Looking at the individual p-values, most variables are statistically significant at conventional levels, including accommodates, bedrooms, bathrooms, beds, minimum_nights, calculated_host_listings_count, number_of_reviews, the shared-room dummy, and all four region dummies. However, not every variable is significant: the hotel-room and private-room dummy variables both have p-values well above 0.05, meaning we cannot confidently say that, holding the other variables fixed, a hotel room or a private room has a different price from the baseline entire-home/apartment category. So while the overall model is informative, a couple of the room-type categories do not show a statistically distinguishable effect on price in this specification.

####3.5 The price column contains some extreme outliers. How much do these outliers affect the fitted model? Retrain your model1 using log(price) as the dependent variable instead of price. Does log-transforming the target reduce the influence of the extreme prices? How do you interpret a coefficient from a model where the dependent variable is log(price) rather than price itself?

```python
# To see how much the most extreme prices are distorting model1, refit the same
# model after removing the most expensive 1% of listings, and compare the fit.
price_99th_percentile = airbnb_clean["price"].quantile(0.99)
airbnb_no_outliers = airbnb_clean[airbnb_clean["price"] <= price_99th_percentile]

predictor_matrix_no_outliers = pd.concat(
    [
        airbnb_no_outliers[numeric_predictors],
        pd.get_dummies(airbnb_no_outliers["room_type"], prefix="room", drop_first=True),
        pd.get_dummies(airbnb_no_outliers["region"], prefix="region", drop_first=True),
    ],
    axis=1,
).astype(float)
X_no_outliers = sm.add_constant(predictor_matrix_no_outliers)
y_no_outliers = airbnb_no_outliers["price"].astype(float)

model1_no_outliers = sm.OLS(y_no_outliers, X_no_outliers).fit()

print(f"R-squared with the top 1% most expensive listings included: {model1.rsquared:.4f}")
print(f"R-squared with the top 1% most expensive listings removed:  {model1_no_outliers.rsquared:.4f}")

print("\nSelected coefficients, with vs. without the most extreme prices:")
comparison = pd.DataFrame({
    "with_outliers": model1.params,
    "without_top_1pct": model1_no_outliers.params,
})
print(comparison.round(2))

# Now retrain model1 using log(price) as the dependent variable
log_price = np.log(airbnb_clean["price"].astype(float))
model1_log = sm.OLS(log_price, X_full).fit()
print(model1_log.summary())

log_predictions = model1_log.predict(X_full)
log_mse = mean_squared_error(log_price, log_predictions)
print(f"\nMean Squared Error of the log-price model (on the log scale): {log_mse:.4f}")
print(f"Adjusted R-squared of the log-price model:                     {model1_log.rsquared_adj:.4f}")
```

The extreme prices have a large effect on the fitted model. When the most expensive 1% of listings are removed, the R-squared of the exact same model more than doubles, rising from roughly 0.26 to roughly 0.53, and several coefficients change substantially in size (for example, the effect of an extra bathroom on price roughly halves). This shows that a handful of very expensive, unusual listings were disproportionately pulling the regression line and inflating the apparent importance of some variables, which is a classic symptom of ordinary least squares being sensitive to outliers because it minimizes squared errors, and squaring a very large error makes it count far more heavily than a small one.

Retraining model1 with log(price) as the dependent variable instead of price produces a dramatically better fit: the adjusted R-squared rises to roughly 0.69, much higher than either of the two models using the raw price. This happens because taking the logarithm compresses the scale of the enormous prices, so a listing priced at ten thousand euros a night no longer dominates the fitting process the way it did on the original scale; the model now treats proportional differences in price as equally important regardless of whether the price itself is small or large, which matches the reality that going from €50 to €100 a night is a similarly meaningful jump as going from €500 to €1000 a night. Log-transforming the target does reduce the influence of extreme prices, because it is movements in percentage terms, not absolute euro terms, that the model is now trying to explain.

The interpretation of a coefficient changes once the dependent variable is log(price). Rather than saying "one more unit of this variable is associated with a change of this many euros in price," each coefficient (when multiplied by 100) is approximately interpreted as a percentage change in price associated with a one-unit increase in the explanatory variable, holding everything else in the model fixed. For example, a bathrooms coefficient of about 0.086 in the log-price model means that, all else equal, one additional bathroom is associated with roughly an 8.6 percent increase in price, rather than a fixed number of extra euros.

###Question 4 - NOAA Storm Events: Classification of property damage

####4.1 Load the dataset and create the target variable named any_damage from the DAMAGE_PROPERTY and DAMAGE_CROPS columns. What proportion of events caused damage?

```python
storm_url = "https://raw.githubusercontent.com/api-sage/diaml_projects/refs/heads/main/project5/StormEvents_details.csv"
storm_raw = pd.read_csv(storm_url, low_memory=False)

# DAMAGE_PROPERTY and DAMAGE_CROPS are stored as text such as "10.00K" or
# "1.50M", meaning thousands or millions of dollars. A missing value means no
# damage amount was reported, which is treated here as zero damage.
def parse_damage_amount(value):
    if pd.isna(value):
        return 0.0
    text = str(value).strip().upper()
    if text == "":
        return 0.0
    multiplier = 1.0
    if text.endswith("K"):
        multiplier, text = 1e3, text[:-1]
    elif text.endswith("M"):
        multiplier, text = 1e6, text[:-1]
    elif text.endswith("B"):
        multiplier, text = 1e9, text[:-1]
    try:
        return float(text) * multiplier
    except ValueError:
        return 0.0

storm_raw["DAMAGE_PROPERTY_USD"] = storm_raw["DAMAGE_PROPERTY"].apply(parse_damage_amount)
storm_raw["DAMAGE_CROPS_USD"] = storm_raw["DAMAGE_CROPS"].apply(parse_damage_amount)

storm_raw["any_damage"] = (
    (storm_raw["DAMAGE_PROPERTY_USD"] > 0) | (storm_raw["DAMAGE_CROPS_USD"] > 0)
).astype(int)

proportion_with_damage = storm_raw["any_damage"].mean()
print(f"Proportion of events that caused some reported property or crop damage: {proportion_with_damage:.2%}")
storm_raw["any_damage"].value_counts()
```

About 30 percent of the storm events in this dataset caused some reported property damage, crop damage, or both, while the remaining 70 percent were recorded with no monetary damage at all. This means the target variable is noticeably imbalanced, with damage-causing events being the minority class, which is an important thing to keep in mind when evaluating a classification model later on, since a model that simply predicted "no damage" for every single event would already be correct about 70 percent of the time without learning anything useful.

####4.3 Compute the duration of each event in minutes, using BEGIN_DATE_TIME and END_DATE_TIME. Also extract the hour of day (0-23) and the day of the week on which each event started from BEGIN_DATE_TIME. Create a new variable named HAS_MAGNITUDE for events that have magnitudes such as wind, hail, etc. Based on data quality, why is it important to use HAS_MAGNITUDE instead of using MAGNITUDE as a predictor in your classification model?

```python
begin_datetime = pd.to_datetime(storm_raw["BEGIN_DATE_TIME"], format="%d-%b-%y %H:%M:%S")
end_datetime = pd.to_datetime(storm_raw["END_DATE_TIME"], format="%d-%b-%y %H:%M:%S")

storm_raw["DURATION_MINUTES"] = (end_datetime - begin_datetime).dt.total_seconds() / 60.0
storm_raw["HOUR"] = begin_datetime.dt.hour
storm_raw["DAY_OF_WEEK"] = begin_datetime.dt.day_name()

# MAGNITUDE records a numeric wind speed or hail size, but it is only reported
# for a handful of event types (mainly wind and hail events) and is missing
# for everything else. HAS_MAGNITUDE simply flags whether a magnitude value
# was reported at all, regardless of its numeric size.
storm_raw["HAS_MAGNITUDE"] = storm_raw["MAGNITUDE"].notna().astype(int)

print(storm_raw[["DURATION_MINUTES", "HOUR"]].describe())
print("\nEvents with a reported magnitude, by event type (top 10):")
print(storm_raw.loc[storm_raw["MAGNITUDE"].notna(), "EVENT_TYPE"].value_counts().head(10))
print(f"\nShare of all events that have a magnitude reported: {storm_raw['HAS_MAGNITUDE'].mean():.2%}")
```

The MAGNITUDE column is missing for well over half of all events, and critically, it is not missing at random: it is essentially only ever recorded for wind-related and hail-related event types, such as thunderstorm wind, hail, high wind, and marine wind events, and it is structurally absent for every other kind of event, such as droughts, floods, or winter storms, because those event types simply do not have a wind-speed or hail-size measurement to report. This means that if MAGNITUDE were used directly as a numeric predictor, its missingness would not reflect a genuine data quality problem to be fixed, but would instead be almost perfectly tied to the event type itself, so the variable would largely be re-encoding information we already have from EVENT_TYPE while introducing a large block of missing values that would otherwise have to be dropped or imputed in a way that distorts the magnitude scale (for example, filling missing magnitudes with zero would wrongly suggest "no wind" for events where wind was never measured in the first place). Using HAS_MAGNITUDE instead avoids this problem by turning a problematic, mostly-missing numeric column into a clean, always-available yes/no flag that captures the useful part of the information, namely whether this type of event comes with an officially measured physical intensity at all, without forcing us to either discard a huge number of rows or inject artificial, misleading numeric values into the model.

####4.4 What is target leakage? Assume the objective is to predict damage using information available at or near the start of the event; explain why INJURIES_DIRECT, INJURIES_INDIRECT, DEATHS_DIRECT, and DEATHS_INDIRECT should not be used as predictors under this goal. Describe a different prediction task in which they would be legitimate predictors.

Target leakage happens when a predictor used to train a model contains information that would not actually be available at the moment the prediction needs to be made in real life, often because that predictor is itself a consequence of the same underlying event being predicted, or because it is only measured or finalized after the outcome is already known. A model trained with a leaky predictor can look deceptively accurate during testing, since the leaked information is highly correlated with the target, but it will fail in real-world use, because the leaked information simply will not exist yet at the time a genuine prediction is needed.

In this task, the goal is explicitly to predict whether an event will cause damage using only information available at or near the start of the event. Injuries and deaths, however, are outcomes that unfold during and after the event itself, in much the same way that property damage does; a storm has to actually happen, and cause its destructive effects, before anyone is injured, killed, or before property gets damaged. In other words, injuries, deaths, and property damage are all downstream consequences of the same underlying severity of the storm, observed and recorded only after the event has already caused its harm. Using the final injury and death counts to predict whether damage occurred would be using information from the future relative to the moment we are supposed to be making the prediction, and it would almost certainly produce an artificially inflated, unrealistic accuracy that could never be reproduced on a truly new, still-unfolding storm event.

These same variables would become legitimate predictors in a different task: for example, if the goal were instead to predict the eventual size of an insurance payout or the total economic cost of damage after a storm has already fully run its course and its human toll is already known. In that kind of retrospective or post-event risk-assessment task, information about how many people were hurt or killed is a perfectly legitimate, already-available input, since the prediction is explicitly being made using the full set of facts about an event that has already finished, rather than trying to forecast the outcome of an event that has only just begun.

####4.6 On the test set from 4.5, report precision, recall, the F1 score, and the ROC AUC, and discuss the results. Why might accuracy alone still give a misleading picture of how well the model performs here? What does each of precision, recall, F1, and ROC AUC tell you that accuracy does not?

On the test set, the model achieves a precision of about 0.61, a recall of about 0.63, an F1 score of about 0.62, and a ROC AUC of about 0.82. Precision tells us that, among all the events the model flagged as likely to cause damage, about 61 percent actually did cause damage, meaning a little under 4 in 10 flagged events are false alarms. Recall tells us that, among all the events that genuinely caused damage, the model successfully caught about 63 percent of them, meaning roughly 37 percent of truly damaging events were missed. The F1 score combines these two numbers into one overall balance measure, giving a single score that penalizes a model which is strong on only one of the two. The ROC AUC of about 0.82 is comfortably above 0.5 (random guessing) and indicates that the model does a good job of ranking events by their underlying probability of damage, correctly placing a randomly chosen damaging event ahead of a randomly chosen non-damaging event about 82 percent of the time, regardless of what specific probability cutoff is used to turn those probabilities into yes/no decisions.

Accuracy alone can give a misleading picture here because the dataset is imbalanced: only about 30 percent of events actually caused damage, as found in question 4.1. A naive model that always predicts "no damage" would already achieve roughly 70 percent accuracy without ever correctly identifying a single damaging event, which would look deceptively good on the accuracy metric alone while being completely useless for the actual goal of flagging dangerous, damage-causing storms. Accuracy treats both kinds of mistakes, wrongly predicting damage and wrongly predicting no damage, as equally costly and simply counts up how often the model is right overall, without revealing anything about how it performs specifically on the minority, damage-causing class that we most care about identifying.

This is exactly what precision, recall, F1, and ROC AUC add on top of accuracy. Precision tells us how trustworthy a positive ("damage") prediction is, which matters if false alarms are costly, for example if an alert triggers an expensive emergency response. Recall tells us how completely the model catches the damaging events we don't want to miss, which matters if failing to detect real damage, for example by under-reporting storms that need urgent follow-up, is the more serious kind of error. The F1 score balances these two competing concerns into a single number when both false alarms and missed detections matter and there is no clearly dominant one. Finally, ROC AUC evaluates the model's ability to separate the two classes across every possible decision threshold, not just the default 0.5 cutoff, giving a threshold-independent sense of how well-calibrated and discriminative the model's underlying probability estimates are. Taken together, these four metrics give a much fuller and more honest picture of performance on an imbalanced classification problem than accuracy can provide on its own.
